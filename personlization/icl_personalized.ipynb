# Blocking a New Category of Harmful Texts(toxic)

## exprimental setup

```python
task_type = 'block_toxic'
category = 'toxic'
new_shots = 0
reasoning = False
random_seed = 42
instruction_type = 'level1-chats'
retrieval = 'random'
shots_list = [0]
harmful = 1
benign = 1
harmful_ratio = harmful/(harmful+benign) # ratio of harmful text samples
dataset = 'textdetox'
json_results = "./results/results_personalization.json"
label_names = ['benign', 'harmful']
# spam
df1 = pd.read_csv('../datasets/spam/sscd.csv', encoding='utf-8', sep='\t')
df1.rename(columns={'v1': 'label', 'v2': 'text'}, inplace=True)
df1['original_label'] = df1['label']
df1['label'] = df1['label'].map({'ham': 'benign','spam': 'harmful'})
# get test set
sms_test_df = df1.sample(frac=0.2, random_state=42)  # random_state 用于确保结果可重复
sms_df_remaining = df1.drop(sms_test_df.index)
sms_df_remaining.reset_index(drop=True, inplace=True)

ham_df_remaining = sms_df_remaining[sms_df_remaining['original_label'] == 'ham']
spam_df_remaining = sms_df_remaining[sms_df_remaining['original_label'] == 'spam']
print(ham_df_remaining.shape)   
print(spam_df_remaining.shape)   

# toxic 
df2 = pd.read_parquet('../datasets/toxic/multilingual_toxicity_dataset/en-00000-of-00001.parquet')
df2.rename(columns={'toxic': 'label'}, inplace=True)
df2['label'] = df2['label'].map({0: 'benign', 1: 'toxic'})
df2['original_label'] = df2['label']
df2['label'] = df2['label'].map({'benign': 'benign', 'toxic': 'harmful'})

# get test set
toxicity_test_df = df2.sample(frac=0.2, random_state=42)  # random_state 用于确保结果可重复
toxicity_df_remaining = df2.drop(toxicity_test_df.index)
toxicity_df_remaining.reset_index(drop=True, inplace=True)

non_toxic_df_remaining = toxicity_df_remaining[toxicity_df_remaining['original_label'] == 'benign']
toxic_df_remaining = toxicity_df_remaining[toxicity_df_remaining['original_label'] == 'toxic']
print(non_toxic_df_remaining.shape)   
print(toxic_df_remaining.shape)   
# emotion
df3 = pd.read_parquet('../datasets/emotion/data/train-00000-of-00001.parquet')
df3.rename(columns={'sentence': 'text'}, inplace=True)
df3['label'] = df3['label'].map({0: 'negative', 1: 'positive'})
df3['original_label'] = df3['label']
df3['label'] = df3['label'].map({'negative': 'harmful', 'positive': 'benign'})

# get test set
emotion_test_df = pd.read_parquet('../datasets/emotion/data/validation-00000-of-00001.parquet')
emotion_test_df.rename(columns={'sentence': 'text'}, inplace=True)
emotion_test_df['label'] = emotion_test_df['label'].map({0: 'negative', 1: 'positive'})
emotion_test_df['original_label'] = emotion_test_df['label']
emotion_test_df['label'] = emotion_test_df['label'].map({'negative': 'harmful', 'positive': 'benign'})

emotion_df_remaining = df3

positive_df_remaining = emotion_df_remaining[emotion_df_remaining['original_label'] == 'positive']
negative_df_remaining = emotion_df_remaining[emotion_df_remaining['original_label'] == 'negative']
print(positive_df_remaining.shape)   
print(negative_df_remaining.shape)   
# merge datasets

test_df = toxicity_test_df.copy().reset_index(drop=True)
# df_remaining = pd.concat([sms_df_remaining, toxicity_df_remaining, emotion_df_remaining], ignore_index=True)
# print(df_remaining.shape)   
# test_df = test_df.head(10)
# test_df.to_csv('./demos.csv', index=False)
```

## run experiments

```python
# model = None
# max_context_length = 32 * 1024
# modelID = 'Qwen2.5-7B-Instruct' # 'Meta-Llama-3.1-8B-Instruct' 'Mistral-7B-Instruct-v0.3'
results = {}
icl = InContextLearner(task_type=task_type, reasoning= reasoning, model_name='', train_df=None, test_df=test_df, model=model, context_length=max_context_length, index_name="")

for n_shots in shots_list:
    for i in [0, 1, 2]:
        random_state = random_seed + i
        
        print(f"n_shots: {n_shots}, random_state: {random_state}, retrieval: {retrieval}, instruction_type: {instruction_type}")
        
        if retrieval != 'random' and i != 0:
            continue
        if new_shots == 0:
            result_file_name = f"./results/confidence/personalization_{task_type}_{category}_{modelID.replace('.','-')}_d{dataset}_i{instruction_type}_r{random_state}_n{n_shots}_r{retrieval}_p{harmful}-{benign}_t{len(test_df)}"
        else:
            result_file_name = f"./results/confidence/personalization_{task_type}_{category}_{modelID.replace('.','-')}_d{dataset}_i{instruction_type}_r{random_state}_n{n_shots}+{new_shots}_r{retrieval}_p{harmful}-{benign}_t{len(test_df)}"
        if reasoning:
            result_file_name += '_reasoning'
        result_file_name += '.csv'
        print(f"result_file_name: {result_file_name}")
        # break
        # result_file_name = './datasets/100+500_predicted.csv'
        if os.path.exists(result_file_name):
            print("This task has been done: result_file_name")
            continue
        
        if retrieval == 'random':
            train_df = pd.concat([ham_df_remaining.sample(n=int(n_shots/4), random_state=random_state), 
                                    spam_df_remaining.sample(n=int(n_shots/4), random_state=random_state), 
                                    toxic_df_remaining.sample(n=int(new_shots), random_state=random_state),
                                    positive_df_remaining.sample(n=int(n_shots/4), random_state=random_state),
                                    negative_df_remaining.sample(n=int(n_shots/4), random_state=random_state)], ignore_index=True)
        else:
            print("non-random retrieval")
            exit(0)
        
        print(train_df.groupby('original_label').size())
        
        index_name = f"{n_shots}-shot-{i}"
        icl.index_name = index_name
        icl.train_df = train_df
        prompts = icl.generate_prompts(system_description=system_descs['personalization'][task_type]
        [instruction_type],n_shots=n_shots+new_shots, retrieval=retrieval)
        # break
        # print(model.get_tokenizer().decode(prompts[0]['prompt_token_ids']))
        # result_df = pd.read_csv(result_file_name)
        # break
        
        outputs = icl.predictGuided(prompts=prompts, label_names=label_names, reasoning=reasoning)
        reasons, predicted_label_list, predicted_labels_list = icl.process_outputs(outputs=outputs, label_names=label_names)        
        # break
        # write the predicted labels and confidences to file
        
            
        result_df = test_df
        result_df['predicted_labels'] = predicted_labels_list
        result_df['predicted_label'] = predicted_label_list
        result_df['reasons'] = reasons
        # result_df['original_outputs'] = outputs
        result_df.to_csv(result_file_name, index=False)
        # break
```
