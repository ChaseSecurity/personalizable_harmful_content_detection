```python
for category in [ 'negative','toxic', 'spam']:
    df = pd.read_csv('../datasets/data_augmentation/' + category + '_augmented.csv')
    train_samples_list = []
    test_samples_list = []
    
    for text in df['text']:
        text_aug = []
        synonym_list, deletion_list, swap_list, insertion_list = augmenter.augment(text)
        for i in range(len(synonym_list)):
            
            text_aug.append(synonym_list[i])
            text_aug.append(deletion_list[i])
            text_aug.append(swap_list[i])
            text_aug.append(insertion_list[i])
        
        if len(text_aug) < 12 or text in text_aug:
            print('error')
        train_samples_list.append(text_aug[:4])
        test_samples_list.append(text_aug[4:])
    df['train_samples_' + str(perturbation_level)] = train_samples_list
    df['test_samples_' + str(perturbation_level)] = test_samples_list
    df.to_csv('../datasets/data_augmentation/' + category + '_augmented.csv', index=False)
```
