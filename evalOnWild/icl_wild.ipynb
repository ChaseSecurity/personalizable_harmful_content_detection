# run experiment

```python
results = {}
icl = InContextLearner(task_type=task_type, reasoning= reasoning, model_name='', train_df=None, test_df=test_df, model=model, context_length=max_context_length, index_name="")

for n_shots in shots_list:
    for i in [0, 1, 2]:
        random_state = random_seed + i
        
        print(f"n_shots: {n_shots}, random_state: {random_state}, retrieval: {retrieval}, instruction_type: {instruction_type}")
        
        if retrieval != 'random' and i != 0:
            continue
        if new_shots == 0:
            result_file_name = f"./results/confidence/{task_type}_{category}_{modelID.replace('.','-')}_d{dataset}_i{instruction_type}_r{random_state}_n{n_shots}_r{retrieval}_p{harmful}-{benign}_t{len(test_df)}"
        else:
            result_file_name = f"./results/confidence/{task_type}_{category}_{modelID.replace('.','-')}_d{dataset}_i{instruction_type}_r{random_state}_n{n_shots}+{new_shots}_r{retrieval}_p{harmful}-{benign}_t{len(test_df)}"
        if reasoning:
            result_file_name += '_reasoning'
        result_file_name += '.csv'
        print(f"result_file_name: {result_file_name}")
        # break
        # result_file_name = './datasets/100+500_predicted.csv'
        if os.path.exists(result_file_name):
            print("This task has been done: result_file_name")
            continue
        
        if retrieval == 'random':
            if 'multi-label' not in dataset:
                train_df = pd.concat([ham_df_remaining.sample(n=int(n_shots/6), random_state=random_state), 
                                    spam_df_remaining.sample(n=int(n_shots/6), random_state=random_state), 
                                    toxic_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    non_toxic_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    positive_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    negative_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    pd.concat([
                                            df_group.sample(n=sample_sizes[label], random_state=random_state)
                                            for label, df_group in new_shots_df.groupby('original_label')
                                            if label in sample_sizes
                                        ])
                                        ], ignore_index=True)
            else:
                train_df = pd.concat([ham_df_remaining.sample(n=int(n_shots/6), random_state=random_state), 
                                    spam_df_remaining.sample(n=int(n_shots/6), random_state=random_state), 
                                    toxic_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    non_toxic_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    positive_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    negative_df_remaining.sample(n=int(n_shots/6), random_state=random_state),
                                    new_shots_df.sample(n=new_shots, random_state=random_state)], ignore_index=True)
        else:
            train_df = df_remaining
        
        print(train_df.groupby('original_label').size())
        
        index_name = f"{n_shots}-shot-{i}"
        icl.index_name = index_name
        icl.train_df = train_df
        prompts = icl.generate_prompts(system_description=system_descs[task_type]
        [instruction_type],n_shots=n_shots+new_shots, retrieval=retrieval)
        # break
        # print(model.get_tokenizer().decode(prompts[0]['prompt_token_ids']))
        # result_df = pd.read_csv(result_file_name)
        # break
        
        outputs = icl.predictGuided(prompts=prompts, label_names=label_names, reasoning=reasoning)
        reasons, predicted_label_list, predicted_labels_list = icl.process_outputs(outputs=outputs, label_names=label_names)        
        # break
        # write the predicted labels and confidences to file
        
            
        result_df = test_df
        result_df['predicted_labels'] = predicted_labels_list
        result_df['predicted_label'] = predicted_label_list
        result_df['reasons'] = reasons
        # result_df['original_outputs'] = outputs
        # result_df.to_csv(result_file_name, index=False)
        result_df.to_csv('result.csv', index=False)
        break
```
